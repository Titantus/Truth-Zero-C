```python
# @title
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import linregress

def plot_transition_law_validation(results_df, P, window_size_fit=50):
    """
    Generates a 6-panel figure to visually compare continuous and discrete
    transition law dynamics, including phase portraits, time evolution,
    error histories, and early-time slope analysis.

    Args:
        results_df (pd.DataFrame): DataFrame containing simulation results.
        P: A parameters object (e.g., from params.py) with attributes like
           theta_0, gamma, Gamma, cycles.
        window_size_fit (int): The window size for the early-time ln(N) fit.
    """
    # --- Core geometric constants (from P object) ---
    theta0 = P.theta_0
    gamma_decay_rate = P.gamma
    P_Gamma_lattice = P.Gamma # Changed from P.gamma_lattice to P.Gamma

    # Extract data from the DataFrame
    t_int = results_df['Cycle'].values
    X_d = results_df['X_discrete'].values
    Y_d = results_df['Y_discrete'].values
    Phi_d = results_df['Phi_discrete'].values
    X_c = results_df['X_continuous'].values
    Y_c = results_df['Y_continuous'].values
    Phi_c = results_df['Phi_continuous'].values

    error_X_hist = results_df['X_error'].values
    error_Y_hist = results_df['Y_error'].values
    error_Phi_hist = results_df['Phi_error'].values

    # Re-calculate RMS errors
    rms_X = np.sqrt(np.mean(error_X_hist**2))
    rms_Y = np.sqrt(np.mean(error_Y_hist**2))
    rms_Phi = np.sqrt(np.mean(error_Phi_hist**2))

    Phi_inf = Phi_d[-1] # Saturated Phi_discrete

    # N_full, Phi_full, lnN_full for Panel 6
    N_full = t_int[1:] # Exclude N=0 for log fitting
    Phi_full = Phi_d[1:]
    lnN_full = np.log(N_full)

    P_cycles = len(t_int) - 1 # P.cycles corresponds to the total number of steps in t_int

    # ============================================================
    # 6-Panel Figure: Transition Law Validation
    # ============================================================

    plt.style.use('dark_background') # Ensure dark background for consistency
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))

    # Define a function for plotting comparisons
    def plot_single_comparison_panel(ax, t_data, y_cont, y_disc, title, ylabel, ccol, dcol, show_legend=False):
        ax.plot(t_data, y_cont, color=ccol, lw=1.5, label='Continuous')
        step = max(1, len(t_data) // 500) # Sample for scatter to avoid clutter
        ax.scatter(t_data[::step], y_disc[::step], color=dcol, marker='x', s=16, label='Discrete', alpha=0.6)
        ax.set_title(title)
        ax.set_ylabel(ylabel)
        ax.grid(True, alpha=0.3)
        if show_legend:
            ax.legend(loc='upper right', fontsize=8)

    # Panel 1: X Continuous vs Discrete
    plot_single_comparison_panel(axes[0, 0], t_int, X_c, X_d,
                                 'X (Travel) Continuous vs Discrete', 'X', 'C0', 'C3', show_legend=True)

    # Panel 2: Y Continuous vs Discrete
    plot_single_comparison_panel(axes[0, 1], t_int, Y_c, Y_d,
                                 'Y (Torque) Continuous vs Discrete', 'Y', 'C2', 'C4', show_legend=True)

    # Panel 3: Φ Continuous vs Discrete
    plot_single_comparison_panel(axes[0, 2], t_int, Phi_c, Phi_d,
                                 r'$\Phi$ (Phase) Continuous vs Discrete', r'$\Phi$', 'C1', 'C5', show_legend=True)

    # Calculate the geometric tail for Panel 4
    phi_geometric_tail = Phi_inf - Phi_d

    # Panel 4: Absolute Error History and Geometric Tail (Combined)
    axes[1, 0].plot(t_int, error_X_hist, label=r'$|X_d - X_c|$', color='C0', alpha=0.7)
    axes[1, 0].plot(t_int, error_Y_hist, label=r'$|Y_d - Y_c|$', color='C2', alpha=0.7)
    axes[1, 0].plot(t_int, error_Phi_hist, label=r'$|\Phi_d - \Phi_c|$', color='C1', alpha=0.7)
    axes[1, 0].plot(t_int, phi_geometric_tail, label=r'$(\Phi_\infty - \Phi_d)$ Geometric Tail', color='C6', alpha=0.7, linestyle='--')
    axes[1, 0].set_title('Absolute Error History and Geometric Tail')
    axes[1, 0].set_ylabel('Absolute Value (Log Scale)')
    axes[1, 0].set_xlabel('Cycle (N)')
    axes[1, 0].set_yscale('log')
    axes[1, 0].legend(fontsize=8)
    axes[1, 0].grid(True, alpha=0.3)

    # Panel 5: Phase Portrait (Y vs X)
    axes[1, 1].plot(X_c, Y_c, color='C0', lw=1.5, label='Continuous')
    step = max(1, len(X_d) // 500)
    axes[1, 1].scatter(X_d[::step], Y_d[::step], color='C3', marker='x', s=16, label='Discrete', alpha=0.6)
    axes[1, 1].set_title('Phase Portrait (Y vs X)')
    axes[1, 1].set_xlabel('X (Travel)')
    axes[1, 1].set_ylabel('Y (Torque)')
    axes[1, 1].grid(True, alpha=0.3)
    axes[1, 1].legend(loc='upper right', fontsize=8)

    # Panel 6: Early-Time ln(N) Fit
    if window_size_fit > 1 and window_size_fit <= len(N_full):
        N_win_current = N_full[:window_size_fit]
        lnN_win_current = lnN_full[:window_size_fit]
        Phi_win_current = Phi_full[:window_size_fit]

        if len(N_win_current) > 1:
            slope_a, intercept_b, r_value, p_value, std_err = linregress(lnN_win_current, Phi_win_current)
            a_over_gamma = slope_a / P_Gamma_lattice

            axes[1, 2].plot(lnN_full, Phi_full, 'o', markersize=3, alpha=0.5,
                            label=r'$\Phi(N)$ data')
            axes[1, 2].plot(lnN_win_current,
                            slope_a * lnN_win_current + intercept_b,
                            color='red', linestyle='--', label='Linear fit')

            axes[1, 2].set_title(r'Early-Time $\Phi(N)$ vs $\ln(N)$ (N='f'{window_size_fit})')
            axes[1, 2].set_xlabel(r'$\ln(N)$')
            axes[1, 2].set_ylabel(r'$\Phi(N)$')
            axes[1, 2].grid(True, alpha=0.3)
            axes[1, 2].legend(loc='upper left', fontsize=8)

            stats_text = (
                f'Slope a: {slope_a:.4e}\n'
                f'Pearson r: {r_value:.3f}\n'
                r'$\Gamma$'+f': {P_Gamma_lattice:.4e}\n'
                r'$a/\Gamma$'+f': {a_over_gamma:.3f}'
            )
            axes[1, 2].text(0.95, 0.05, stats_text, transform=axes[1, 2].transAxes,
                            fontsize=8, va='bottom', ha='right',
                            bbox=dict(boxstyle="round,pad=0.3", fc="white",
                                      ec="gray", lw=0.5, alpha=0.7))
        else:
            axes[1, 2].text(0.5, 0.5, 'Not enough data for early-time fit',
                            ha='center', va='center', transform=axes[1, 2].transAxes)
    else:
        axes[1, 2].text(0.5, 0.5, 'Window size invalid for early-time fit',
                        ha='center', va='center', transform=axes[1, 2].transAxes)

    # Overall figure title and subtitle
    fig.suptitle('Transition Law Validation — Continuous vs Discrete Dynamics',
                 fontsize=18, y=1.02, x=0.5)

    fig.text(0.5, 0.98,
             r'$\theta_0$'+f'={np.degrees(theta0):.4f} deg | '
             r'$\gamma$'+f'={gamma_decay_rate:.4f} | '
             r'$\Gamma$'+f'={P_Gamma_lattice:.4e} | '
             f'Cycles={P_cycles}',
             fontsize=10, ha='center', va='bottom', color='gray')

    # Figure footer with RMS errors
    footer_text = (
        f'RMS(X)={rms_X:.2e} | '
        f'RMS(Y)={rms_Y:.2e} | '
        r'RMS($\Phi$)'+f'={rms_Phi:.2e}'
    )
    fig.text(0.5, 0.005, footer_text, ha='center', va='bottom',
             fontsize=10, color='lightgray')

    plt.tight_layout(rect=[0.01, 0.04, 0.99, 0.95])
    plt.show()

    ## 2. Analytical Verification
    ### 3a. Simulation Kernels (Functions)

    import pandas as pd
import numpy as np
from scipy.integrate import odeint

def run_transition_law_simulation(P, num_cycles=None, X0=1.0, Y0=0.0, Phi0=0.0):
    """
    Consolidated function to run both discrete and continuous transition law simulations.

    Args:
        P: A parameters object (e.g., from params.py) containing:
           - P.theta_0: Rotation angle for discrete, angular frequency for continuous.
           - P.gamma: Decay rate.
           - P.sigma: Saturation parameter.
           - P.cycles: Total number of cycles for simulation (if num_cycles is None).
        num_cycles: Override P.cycles for custom simulation length.
        X0, Y0, Phi0: Initial conditions for X, Y, Phi.

    Returns:
        pd.DataFrame: A DataFrame containing discrete, continuous, and error metrics.
    """
    total_cycles = num_cycles if num_cycles is not None else P.cycles

    # Ensure P.sigma exists, provide a default if not (though it should be derived)
    sigma_sq = P.sigma**2 if hasattr(P, 'sigma') and P.sigma is not None else 1.0

    # --- Discrete Simulation ---
    t_disc = np.arange(0, total_cycles + 1)
    X_d = np.zeros(total_cycles + 1)
    Y_d = np.zeros(total_cycles + 1)
    Phi_d = np.zeros(total_cycles + 1)

    X_d[0] = X0
    Y_d[0] = Y0
    Phi_d[0] = Phi0

    for i in range(total_cycles):
        current_amplitude_sq = X_d[i]**2 + Y_d[i]**2
        w = 1.0 / (1.0 + current_amplitude_sq / sigma_sq) if sigma_sq != 0 else 1.0

        # Apply rotation (P.theta_0 is the angle)
        X_rotated = X_d[i] * np.cos(P.theta_0) - Y_d[i] * np.sin(P.theta_0)
        Y_rotated = X_d[i] * np.sin(P.theta_0) + Y_d[i] * np.cos(P.theta_0)

        # Apply decay and saturation
        X_d[i+1] = (1.0 - P.gamma) * w * X_rotated
        Y_d[i+1] = (1.0 - P.gamma) * w * Y_rotated
        Phi_d[i+1] = Phi_d[i] + P.gamma * current_amplitude_sq

    # --- Continuous Simulation ---
    # Define the ODE system
    def continuous_equations(state, t_val, params):
        X_c_eq, Y_c_eq, Phi_c_eq = state
        theta_0_cont = params.theta_0 # Treat as angular frequency for continuous
        gamma_cont = params.gamma
        sigma_sq_cont = params.sigma**2 if hasattr(params, 'sigma') and params.sigma is not None else 1.0

        current_amplitude_sq_cont = X_c_eq**2 + Y_c_eq**2
        w_cont = 1.0 / (1.0 + current_amplitude_sq_cont / sigma_sq_cont) if sigma_sq_cont != 0 else 1.0

        # Equations based on damped oscillator with rotation and saturation
        dXdt = -gamma_cont * X_c_eq + w_cont * (-theta_0_cont * Y_c_eq)
        dYdt = -gamma_cont * Y_c_eq + w_cont * (theta_0_cont * X_c_eq)
        dPhidt = gamma_cont * current_amplitude_sq_cont # Phi accumulation

        return [dXdt, dYdt, dPhidt]

    t_cont = np.linspace(0, total_cycles, total_cycles + 1) # Match discrete time points
    initial_state_cont = [X0, Y0, Phi0]
    solution_cont = odeint(continuous_equations, initial_state_cont, t_cont, args=(P,))

    X_c = solution_cont[:, 0]
    Y_c = solution_cont[:, 1]
    Phi_c = solution_cont[:, 2]

    # --- Calculate Errors ---
    # Errors are absolute differences at each point
    error_X_hist = np.abs(X_d - X_c)
    error_Y_hist = np.abs(Y_d - Y_c)
    error_Phi_hist = np.abs(Phi_d - Phi_c)

    # --- Store Results in DataFrame ---
    results_df = pd.DataFrame({
        'Cycle': t_disc, # Use discrete cycles as the common time index
        'X_discrete': X_d,
        'Y_discrete': Y_d,
        'Phi_discrete': Phi_d,
        'X_continuous': X_c,
        'Y_continuous': Y_c,
        'Phi_continuous': Phi_c,
        'X_error': error_X_hist,
        'Y_error': error_Y_hist,
        'Phi_error': error_Phi_hist,
    })

    csv_filename = "transition_law_validation_results.csv"
    results_df.to_csv(csv_filename, index=False)
    print(f"Simulation results saved to {csv_filename}")

    return results_df
    ### 3b. Sigma² Candidate Testing
```

```python
# @title
import numpy as np
from dataclasses import dataclass
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

# ----------------------------------------------------------
# 1. Unified Parameters (using P for common constants)
# ----------------------------------------------------------

# Slip-Pivot specific constants
eta: float = (1/3) * (2 / np.pi)            # = 2/(3π)

# Lattice-strain mechanical model constants
mass: float = 18.0
radius: float = 0.165
shaft_len: float = 0.91
gravity: float = 9.81

# Simulation sizes
slip_steps: int = 2000
sync_steps: int = 500
rpm_steps: int = 1000

# ----------------------------------------------------------
# 2. Slip-Pivot Operator + Discrete Map
# ----------------------------------------------------------
class SlipPivotSimulator:
    def __init__(self, gamma, theta_0, sigma, eta):
        self.gamma = gamma
        self.sigma_sq = sigma**2 # Use sigma squared directly
        self.S = np.array([[1 - eta, eta],
                           [eta, 1 - eta]])
        self.R = np.array([[np.cos(theta_0), -np.sin(theta_0)],
                           [np.sin(theta_0),  np.cos(theta_0)]])

    def run(self, N=2000, X0=1.0, Y0=0.0):
        state = np.array([X0, Y0], dtype=float)
        Phi = 0.0
        for _ in range(N):
            A = state @ state
            w = 1.0 / (1.0 + A / self.sigma_sq)
            state = self.R @ state
            state = self.S @ state
            state = (1.0 - self.gamma) * w * state
            Phi += self.gamma * A
        return Phi

sim = SlipPivotSimulator(P.gamma, P.theta_0, P.sigma, eta)
phi_slip = sim.run(slip_steps)
baseline = 0.04047712 # This baseline is hardcoded, consider if it should be derived or in P
dev_pct = (phi_slip - baseline) / baseline * 100

# ----------------------------------------------------------
# 3. Rate-Independent Sync-Cost (abstract slider / phasor)
# ----------------------------------------------------------
def rate_independent_sync(n_steps=500, K=1.0):
    slider = np.linspace(0, 4*np.pi, n_steps)
    incoming = slider + np.pi/3
    dtheta = incoming - slider
    F_sync = K * np.sin(dtheta)
    mass_res = np.abs(F_sync)
    heat = 1.0 - np.abs(np.cos(dtheta))
    return dtheta, mass_res, heat

delta, mass_res, heat = rate_independent_sync(sync_steps)

# ----------------------------------------------------------
# 4. Lattice-Strain Reallocation (spinning shell → precession)
# ----------------------------------------------------------
I = 0.5 * mass * radius**2
tau_static = shaft_len * mass * gravity

def lattice_strain(rpm_array, I, tau_static):
    precession = np.zeros_like(rpm_array)
    strain = np.zeros_like(rpm_array)
    for i, rpm in enumerate(rpm_array):
        omega = rpm * 2 * np.pi / 60.0
        if omega < 1e-12:
            precession[i] = 0.0
            strain[i] = tau_static
        else:
            L = I * omega
            precession[i] = tau_static / L
            # Bounded fall-off that matches the conceptual plot
            strain[i] = tau_static / (1.0 + precession[i] * omega * 0.002)
    return precession, strain

rpm = np.linspace(0, 2500, rpm_steps)
precession, strain = lattice_strain(rpm, I, tau_static)

# ----------------------------------------------------------
# 5. Multi-panel figure suite
# ----------------------------------------------------------
plt.style.use('dark_background')
fig, axes = plt.subplots(1, 2, figsize=(13.5, 5.2), constrained_layout=True)

# --- Left: Lattice Strain ---
ax1 = axes[0]
ax1.plot(rpm, strain, color='#00E5FF', lw=2.2)
ax1.axhline(0, color='gray', lw=0.6, alpha=0.5)
ax1.set_xlim(0, 2500)
ax1.set_ylim(-5, tau_static * 1.08)
ax1.set_xlabel('Wheel / Shell Spin (RPM)')
ax1.set_ylabel('Downward Lattice Strain (N·m)', color='#00E5FF')
ax1.tick_params(axis='y', labelcolor='#00E5FF')
ax1.set_title('Lattice Strain Reallocation\n(Spinning shell off-loads downward load into precession)',
              fontsize=10)

ax1.annotate(
    "External baseline pressure (≈1)\nwins at 33rd channel\n→ planar decoherence\n→ heat-trapping bubble",
    xy=(180, 40), xytext=(620, 95),
    fontsize=8, color='#FFAB40',
    arrowprops=dict(arrowstyle='->', color='#FFAB40', lw=1.2),
    bbox=dict(boxstyle="round,pad=0.35", fc="#121212",
              ec="#FFAB40", lw=0.5, alpha=0.9)
)

# --- Right: Rate-Independent Sync-Cost ---
ax2 = axes[1]
ax2.plot(delta, mass_res, color='#00E5FF', lw=1.9, label='Mass resistance  |F_sync|')
ax2.plot(delta, heat, color='#FF6E40', lw=1.9, alpha=0.9, label='Heat dissipation  1−|cos Δθ|')
ax2.axhline(0, color='gray', lw=0.6, alpha=0.5)
ax2.set_xlabel('Phase misalignment Δθ (rad)')
ax2.set_ylabel('Cost (arb. units)')
ax2.set_title('Rate-Independent Sync-Cost\n(Abstract slider / phasor realignment)', fontsize=10)
ax2.legend(loc='upper right', fontsize=8.5, framealpha=0.25)
ax2.set_xlim(delta.min(), delta.max())

fig.suptitle(
    "T′Z₀C  ·  Shell/Octant Asymmetry  ·  Mass = realignment cost  ·  Heat = forced non-alignment cost",
    fontsize=11, color='#E0E0E0', y=1.03
)
plt.show()

# ----------------------------------------------------------
# 6. Consolidated summary table (CSV-ready)
# ----------------------------------------------------------
summary = pd.DataFrame([
    # Slip-Pivot
    ['Slip-Pivot', 'Φ_∞ (with S)', f'{phi_slip:.8f}'],
    ['Slip-Pivot', 'Baseline Φ_∞', f'{baseline:.8f}'],
    ['Slip-Pivot', 'Deviation', f'{dev_pct:+.2f} %'],
    ['Slip-Pivot', 'η = 2/(3π)', f'{eta:.8f}'],
    ['Slip-Pivot', 'λ₋ = 1−2η', f'{1-2*eta:.8f}'],

    # Rate-Independent Sync-Cost
    ['Sync-Cost', 'Steps', sync_steps],
    ['Sync-Cost', 'Mean |Δθ| (rad)', f'{np.mean(np.abs(delta)):.4f}'],
    ['Sync-Cost', 'Mean mass resistance', f'{np.mean(mass_res):.4f}'],
    ['Sync-Cost', 'Mean heat cost', f'{np.mean(heat):.4f}'],

    # Lattice Strain
    ['Lattice-Strain', 'Static torque (0 RPM)', f'{tau_static:.2f} N·m'],
    ['Lattice-Strain', 'I (moment of inertia)', f'{I:.5f} kg·m²'],
    ['Lattice-Strain', 'RPM sweep points', rpm_steps],
])

summary.columns = ['Module', 'Metric', 'Value']

print('\n' + '='*64)
print('CONSOLIDATED SUMMARY')
print('='*64)
display(summary.style.hide(axis='index').set_properties(**{
    'text-align': 'left',
    'font-family': 'monospace',
    'font-size': '12px'
}))
```

```python
# @title
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import os

# The previous os.chdir('/home/workdir/artifacts') line was removed as the directory does not exist.
# Files will be saved to the default working directory.

np.random.seed(42)

# ============================================================
# Discrete Lattice Monte Carlo: 16-state ledger + Axiom 0.1
# ============================================================

N_SECTORS = 8          # spatial octants
N_POLARITIES = 2       # Straight / Loop (Axiom 0.1)
N_STATES = N_SECTORS * N_POLARITIES   # 16
SIN_THETA0 = 1.0 / 3.0
BUFFON_P = 2.0 / np.pi
N_TRIALS = 20000
N_STEPS = 500          # continuous routing steps per trial

print('=== 16-State Ledger Monte Carlo ===')
print(f'Sectors: {N_SECTORS}, Polarities: {N_POLARITIES}, States: {N_STATES}')
print(f'sin θ0 = {SIN_THETA0:.6f}, Buffon p = {BUFFON_P:.6f}')
print()

# ------------------------------------------------------------
# 1. Pure combinatorial occupancy under continuous routing
#    A process must keep exactly one polarity active in each
#    visited sector (Axiom 0.1 mutual exclusivity). Residual
#    is the fraction of the ledger that cannot be cancelled.
# ------------------------------------------------------------

def continuous_routing_trial(n_steps=N_STEPS):
    # occupancy vector over 16 states (8 sectors × 2 polarities)
    occ = np.zeros(N_STATES, dtype=float)
    # start with a random active state (must occupy something)
    active = np.random.randint(0, N_STATES)
    occ[active] = 1.0
    residual_history = []
    for t in range(n_steps):
        sector = active // N_POLARITIES
        pol = active % N_POLARITIES
        # Axiom 0.1: cannot be in both polarities of same sector
        # force switch of polarity or move to adjacent sector
        if np.random.rand() < 0.5:
            # flip polarity in same sector
            new_pol = 1 - pol
            active = sector * N_POLARITIES + new_pol
        else:
            # move to a random other sector, keep polarity or flip
            new_sector = np.random.randint(0, N_SECTORS)
            new_pol = np.random.randint(0, N_POLARITIES)
            active = new_sector * N_POLARITIES + new_pol
        # continuous occupation: add unit weight to current state
        # then renormalise so total weight stays 1 (normalised process)
        occ *= 0.98   # mild decay / reset of old hysteresis
        occ[active] += 1.0
        occ /= occ.sum()
        # residual = min occupation among currently non-zero states
        nonzero = occ[occ > 1e-12]
        residual_history.append(nonzero.min() if len(nonzero) else 0.0)
    return np.array(residual_history), occ

# Run many trials
min_residuals = []
final_occupancies = []
for i in range(N_TRIALS):
    hist, final_occ = continuous_routing_trial()
    min_residuals.append(hist.min())
    final_occupancies.append(final_occ)

min_residuals = np.array(min_residuals)
final_occupancies = np.array(final_occupancies)

print('--- Continuous routing (normalised process) ---')
print(f'Mean of per-trial minimum residual: {min_residuals.mean():.6f}')
print(f'Median of per-trial minimum residual: {np.median(min_residuals):.6f}')
print(f'5th–95th percentile: {np.percentile(min_residuals,5):.6f} – {np.percentile(min_residuals,95):.6f}')
print(f'Target 1/16 = {1/16:.6f}')
print()

# Average final occupancy distribution
mean_occ = final_occupancies.mean(axis=0)
print('Mean final occupancy per state (should be ~1/16 if uniform):')
print(np.round(mean_occ, 4))
print(f'Sum check: {mean_occ.sum():.6f}')
print()

# ------------------------------------------------------------
# 2. Buffon-weighted residual on the same 16-state ledger
# ------------------------------------------------------------
def buffon_residual_trial():
    # each of 16 states is occupied with probability Buffon_p independently
    # then residual floor is the expected min positive weight after
    # mutual-exclusivity projection (cannot occupy both polarities of a sector)
    raw = (np.random.rand(N_STATES) < BUFFON_P).astype(float)
    # enforce Axiom 0.1 per sector: keep at most one polarity
    for s in range(N_SECTORS):
        pair = raw[s*2:(s+1)*2]
        if pair.sum() > 1:
            # randomly keep one
            keep = np.random.randint(0,2)
            pair[:] = 0
            pair[keep] = 1
            raw[s*2:(s+1)*2] = pair
    if raw.sum() == 0:
        return 0.0, raw
    raw /= raw.sum()   # normalise
    nonzero = raw[raw > 0]
    return nonzero.min(), raw

buffon_mins = []
for i in range(N_TRIALS):
    m, _ = buffon_residual_trial()
    buffon_mins.append(m)
buffon_mins = np.array(buffon_mins)

print('--- Buffon-weighted residual (after Axiom 0.1 projection) ---')
print(f'Mean min residual: {buffon_mins.mean():.6f}')
print(f'Median min residual: {np.median(buffon_mins):.6f}')
print(f'5th–95th: {np.percentile(buffon_mins,5):.6f} – {np.percentile(buffon_mins,95):.6f}')
print()

# ------------------------------------------------------------
# 3. Apply diode shear to obtain candidate D_T,min
# ------------------------------------------------------------
# After obtaining a 1/16-scale floor, multiply by sin θ0
shear_floor = (1.0/16.0) * SIN_THETA0
print('--- Diode-shear thermal floor ---')
print(f'(1/16)*sinθ0 = {shear_floor:.6f}  (target 1/48 = {1/48:.6f})')
print(f'Empirical mean continuous-routing min * sinθ0 = {min_residuals.mean()*SIN_THETA0:.6f}')
print(f'Empirical mean Buffon min * sinθ0 = {buffon_mins.mean()*SIN_THETA0:.6f}')
print()

# ------------------------------------------------------------
# 4. Propagation efficiency 1 - D_T,min
# ------------------------------------------------------------
print('--- Propagation limit ---')
print(f'1 - 1/48 = {1 - 1/48:.6f}  →  v_max ≈ {47/48:.6f} c')
print()

# ------------------------------------------------------------
# 5. Quick diagnostic: distribution of number of active states
# ------------------------------------------------------------
n_active = (final_occupancies > 1e-8).sum(axis=1)
print('--- Active-state count at end of continuous routing ---')
print(f'Mean number of active states: {n_active.mean():.2f}')
print(f'Median: {np.median(n_active):.1f}')
print(f'Histogram (0-16):', np.bincount(n_active, minlength=17))
print()

# Save a simple plot of residual histogram
fig, axes = plt.subplots(1,2, figsize=(10,4))
axes[0].hist(min_residuals, bins=50, density=True, alpha=0.7, label='continuous routing')
axes[0].axvline(1/16, color='r', ls='--', label='1/16')
axes[0].set_xlabel('minimum residual')
axes[0].set_ylabel('density')
axes[0].legend()
axes[0].set_title('Continuous-routing min residual')

axes[1].hist(buffon_mins, bins=50, density=True, alpha=0.7, color='orange', label='Buffon + Axiom 0.1')
axes[1].axvline(1/16, color='r', ls='--', label='1/16')
axes[1].set_xlabel('minimum residual')
axes[1].legend()
axes[1].set_title('Buffon-weighted min residual')
plt.tight_layout()
plt.savefig('16state_montecarlo_residuals.png', dpi=120)
print('Plot saved: 16state_montecarlo_residuals.png')
```

```python
# @title
import numpy as np
import matplotlib.pyplot as plt

def simulate_water_phases(stress_range, trials=200):
    """
    Simulates Torque Routing Coherence (eta_tau) across a 32-DOF ledger
    for water under increasing Straight-Mode thermal stress.
    The phase transitions are now emergent from a continuous binding
    gradient and cumulative shear pressure.
    """
    # T'Z0C Canvas Primitives
    P_PI = 2.0 / np.pi               # ~0.6366 Background noise floor
    LATTICE_VARIANCE = P_PI * (1.0 - P_PI) # ~0.2313 Base structural noise
    DOF = 32                         # 8 octants x 4 phases ledger

    # Water-Specific Geometric Calibration
    # 109.47 (Ideal Tetrahedral) vs 104.5 (Actual Water Bond)
    geometric_frustration = (109.47 - 104.5) / 109.47

    # Binding Energy Constants (from v6.5 Item 47, in GPa)
    U_MELT = 0.31    # Outer lanes (Solid -> Liquid slip)
    U_VAP = 2.26     # Mid lanes (Liquid -> Gas slip)
    U_ION = 67.0     # Inner core lanes (Gas -> Plasma slip)

    # --- New: Continuous Binding Energy Gradient ---
    # Distribute binding energy linearly from U_ION (innermost, index 0) to U_MELT (outermost, index DOF-1)
    U_binding_gradient = np.linspace(U_ION, U_MELT, DOF)

    # --- New: Differential Velocity and Shear Heat ---
    # Assume lane index 'i' (0 to 31) represents increasing radius 'r'
    # V_max_scale is arbitrary, determines the magnitude of shear heat.
    V_max_scale = 10.0 # Arbitrary scaling for velocity, tuned for emergent pressure
    lane_velocities = np.array([V_max_scale / (i + 1) for i in range(DOF)])

    # Calculate shear heat between adjacent lanes (proportional to (delta_v)^2)
    # The innermost lane (index 0) has no 'inner' neighbor for shear calculation, so we handle it.
    shear_heats = np.zeros(DOF)
    for i in range(DOF - 1):
        shear_heats[i] = (lane_velocities[i] - lane_velocities[i+1])**2
    # For the outermost lane, assume it shears against a fixed boundary or similar.
    # For simplicity, we can set shear_heats[DOF-1] = shear_heats[DOF-2] or similar, or 0.
    # Given 'cumulative pressure from the core and pushes outward', we'll make shear_heats[i]
    # be the shear *generated by* lane i due to its relative motion with i+1 (or some other convention).
    # Let's consider shear_heats[i] as the shear between lane i and i+1.
    # If the last lane is the outermost, it has no outer neighbor for this calculation in the same way.
    # For cumulative sum, let's say shear_heats[i] is between i and i+1, so cumulative pressure at lane k is sum up to shear_heats[k-1].

    # Cumulative pressure pushing outward. The shear heat generated by inner lanes reduces effective binding of outer lanes.
    cumulative_shear_pressure = np.cumsum(shear_heats) # This array will have DOF elements
    # The last element of cumulative_shear_pressure is the total shear from 0 to DOF-2.
    # We need to map this correctly. If cumulative_shear_pressure[k] is sum from lane 0 to k,
    # then for lane `i`, the cumulative pressure pushing from inside is cumulative_shear_pressure[i-1].
    # For lane 0, this should be 0.
    cumulative_pressure_effect = np.concatenate(([0.0], cumulative_shear_pressure[:-1]))

    # Pressure effect factor to scale the cumulative shear pressure to a relevant GPa range
    # This factor needs to be tuned so that the effect is meaningful for phase transitions.
    pressure_effect_factor = 0.05 # Tunable parameter

    coherence_results = []

    for T in stress_range:
        locked_lanes_total = 0

        for _ in range(trials):
            # 1. Background Noise Floor (The 2/pi "hum")
            lattice_noise = np.random.normal(P_PI, np.sqrt(LATTICE_VARIANCE), DOF)

            # 2. Thermal Vector Stress (Straight-Mode Push)
            # Total stress = Thermal energy (T) + Baseline noise + Geometric pre-load
            lane_stress = (T * (1.0 + geometric_frustration)) + lattice_noise

            # 3. Evaluate 32-DOF Ledger Coherence (Loop-Mode Lock)
            locked_lanes = 0
            for i in range(DOF):
                # Effective binding is reduced by the cumulative internal pressure
                effective_binding = U_binding_gradient[i] - (cumulative_pressure_effect[i] * pressure_effect_factor)

                # A lane is locked if the applied stress is less than its effective binding energy
                if lane_stress[i] < effective_binding:
                    locked_lanes += 1

            locked_lanes_total += locked_lanes

        # Torque Routing Coherence (eta_tau) = Ratio of locked lanes to total DOF
        eta_tau = (locked_lanes_total / trials) / DOF
        coherence_results.append(eta_tau)

    return coherence_results

# Execute Sweep (0 to 100 GPa equivalent thermal stress)
stress_sweep = np.linspace(0, 100, 1000)
coherence_curve = simulate_water_phases(stress_sweep)

# Visualization
plt.figure(figsize=(12, 6))
plt.plot(stress_sweep, coherence_curve, color='navy', linewidth=2.5, label=r'Torque Routing Coherence ($\eta_\tau$)')

# Phase Threshold Markers (as conceptual reference, not directly from simulation logic now)
plt.axvline(x=0.31, color='lightblue', linestyle='--', label=r'U_MELT (0.31 GPa)')
plt.axvline(x=2.26, color='blue', linestyle='--', label=r'U_VAP (2.26 GPa)')
plt.axvline(x=67.0, color='red', linestyle='--', label=r'U_ION (67.0 GPa)')

# Annotations & Formatting
plt.title(r"T'Z0C Scale-Invariant Ledger: Water Phase Transitions (Emergent)", fontsize=14, fontweight='bold')
plt.xlabel(r"Straight-Mode Thermal Stress (Equivalent GPa)", fontsize=12)
plt.ylabel(r"Octant Torque Coherence ($\eta_\tau$)", fontsize=12)
plt.fill_between(stress_sweep, coherence_curve, alpha=0.1, color='navy')
plt.grid(True, alpha=0.3)
plt.legend(loc='upper right', fontsize=10)
plt.xlim(0, 100)
plt.ylim(0, 1.05)
plt.tight_layout()

plt.show()
```
